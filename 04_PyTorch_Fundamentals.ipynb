####**More precise tensore creation**

```python
float_32_tensor = torch.tensor([3.2, 6.1, 9.7],
                               dtype=torch.float32,
                               device="cpu",
                               requires_grad=False)

print("shape: ", float_32_tensor.shape)
print("Dimension: ", float_32_tensor.ndim)
print("dtype: ", float_32_tensor.dtype)
print("device: ", float_32_tensor.device)
```

####**to()**

```python
new_tensor = float_32_tensor.to(torch.int8)
print(new_tensor)
print("dtype: ", new_tensor.dtype)
```

```python
# stack to()
new_tensor.to(torch.float16).to('cpu')
```

####**moving tensor to GPU**

```python
tensor = torch.tensor([1, 2, 3])

print(tensor, tensor.device)
```

```python
torch.cuda.is_available()
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'

tensor_on_gpu = tensor.to(device)
tensor_on_gpu.device
```

####**Moving tensor back to the cpu**

```python
try:
  tensor = tensor_on_gpu.numpy()
except Exception as err:
  print("Error: ", err)
```

```python
tensor_back_on_cpu = tensor_on_gpu.cpu().numpy()
tensor_back_on_cpu
```

Alternative ways to convert data to tensor

```python
import numpy as np
np_array = np.array([1,2,3,4], dtype=np.float64)
```

torch.from_numpy

```python
tensor_from_numpy = torch.from_numpy(np_array)

print(np_array)
print(tensor_from_numpy)
```

```python
np_array[0] = 100
print(np_array)
print(tensor_from_numpy)
```

torch.as_tensor

```python
np_array = np.array([1,2,3,4,5], dtype=np.float64)

tensor_as_tensor = torch.as_tensor(np_array)

print(np_array)
print(tensor_as_tensor)
```

```python
np_array[1]=20

print(np_array)
print(tensor_as_tensor)
```

```python
input_tensor = torch.arange(0,5, dtype=torch.int16)
print(input_tensor)
tensor_as_torch = torch.as_tensor(input_tensor)
print(tensor_as_torch)

tensor_as_torch[-2] = 20
print("input Tensor: ", input_tensor)
print("tensor_as_torch: ", tensor_as_torch)
```

```python
input_tensor = torch.arange(0,5, dtype=torch.int16)
tensor_as_torch = torch.as_tensor(input_tensor, dtype=torch.float64)


tensor_as_torch[-2] = 20
print("input Tensor: ", input_tensor)
print("tensor_as_torch: ", tensor_as_torch)
```

```python
list_data = [1,2,3,4,5]
tensor = torch.as_tensor(list_data)
print(tensor)

tensor[0]= 33
print(list_data)
print(tensor_as_torch)
```

```python
np_array = np.array([1,2,3,4,5])
tensor_np_array = torch.tensor(np_array)

list_data = [1,2,3,4,5]
tensor_from_list = torch.tensor(list_data)

print(tensor_np_array)
print(tensor_from_list)
```

```python
tensor_np_array[0] = 100
print(tensor_np_array)
print(tensor_from_list)
```
